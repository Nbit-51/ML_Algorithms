```python
import sympy as sp
x=sp.Symbol("x")
y=sp.Symbol("y")
f=x*y
df=sp.diff(f,y)
print(df)
```

```python
f=3*x*2+2*x+5
import scipy
from scipy.differentiate import derivative
def f(x):
  return 3*x*2+2*x+5
result=derivative(f,2)
print("Derivative",result.df)
print("Estimated error",result.error)
```

```python
import numpy as np
from sklearn.decomposition import PCA
X=np.array([[1,2],[2,4],[3,6],[4,8]])
pca=PCA(n_components=1)
X_reduced=pca.fit_transform(X)
print(X_reduced)
```

```python
Y=np.array([[1,2,3],[2,4,5],[3,5,7],[4,7,9],[5,8,11]])
Y_mean=np.mean(Y,axis=0)
Y_centered=np.subtract(Y,Y_mean)
Covar=Y_centered.T@Y_centered/(Y_centered.shape[0]-1)
x=np.linalg.eigh(Covar)
print(x.eigenvectors)
W=x.eigenvectors[:,-2:]
W.shape
Y_pca=Y_centered@W
print(Y_pca.shape)
selected=x.eigenvalues[-2:]
sel_sum=np.sum(selected)
perc=sel_sum/np.sum(x.eigenvalues)
print(perc*100)
```
